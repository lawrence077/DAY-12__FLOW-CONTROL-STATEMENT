```python
for i in range(0, 10):
    print(i)
    if i==8:
        break
```

```python
for i in range(0, 10):
    if i==8:
        continue
    print(i)
```

```python
for i in range(0, 10):
    if i==8:
        continue
    print(i, end=" ")
```
